###**Handle Missing Values**

```python
mode_value = df["When do you prefer to study?"].mode()[0]

print("Most common study time:", mode_value)
df["When do you prefer to study?"] = df[
    "When do you prefer to study?"
].fillna(mode_value)
df.isnull().sum()
```

###**Age Analysis**

```python
age_counts = df["What is your age?"].value_counts()

print(age_counts)
```

###**Age Bar Chart**

```python
plt.figure(figsize=(8,5))

plt.bar(
    age_counts.index,
    age_counts.values
)

plt.title("Distribution of Age")
plt.xlabel("Age Groups")
plt.ylabel("Number of Students")

plt.show()
```

###**Age Pie Chart**

```python
plt.figure(figsize=(7,7))

plt.pie(
    age_counts,
    labels=age_counts.index,
    autopct="%1.1f%%"
)

plt.title("Distribution of Age")

plt.show()
```

###**Daily Screen Time Analysis**

```python
screen_time_counts = df[
    "How many hours do you spend on screens daily? (including phone, laptop, tablets )"
].value_counts()

plt.figure(figsize=(9,5))

plt.bar(
    screen_time_counts.index,
    screen_time_counts.values
)

plt.title("Daily Screen Time")
plt.xlabel("Hours")
plt.ylabel("Number of Students")

plt.xticks(rotation=45)

plt.show()
```

###**Bivariate Analysis**

###**Age vs Study Time**

```python
plt.figure(figsize=(9,5))

sns.countplot(
    x="What is your age?",
    hue="When do you prefer to study?",
    data=df
)

plt.title("Age vs Preferred Study Time")
plt.tight_layout()

plt.show()
```

###**Age vs Exercise**

```python
plt.figure(figsize=(9,5))

sns.countplot(
    x="What is your age?",
    hue="How often do you exercise in a week?",
    data=df
)

plt.title("Age vs Exercise Frequency")
plt.tight_layout()

plt.show()
```

###**Screen Time vs Age**

```python
plt.figure(figsize=(10,5))

sns.countplot(
    x="How many hours do you spend on screens daily? (including phone, laptop, tablets )",
    hue="What is your age?",
    data=df
)

plt.title("Screen Time vs Age")
plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

###**Coffee/Tea vs Languages**

###**Commute vs Exercise**

```python
plt.figure(figsize=(10,5))

sns.countplot(
    x="How do you primarily commute to work or school?",
    hue="How often do you exercise in a week?",
    data=df
)

plt.title("Commute Method vs Exercise Frequency")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

###**Sleep vs Screen Time**
